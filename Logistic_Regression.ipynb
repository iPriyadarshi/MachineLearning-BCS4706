```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/iPriyadarshi/MachineLearning-BCS4706/main/dataset/Purchase_Logistic.csv"
purchasedata = pd.read_csv(url)
```

```python
purchasedata.head()
```

```python
X = purchasedata.iloc[:,[2,3]].values
X
```

```python
y = purchasedata.iloc[:,4].values
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
```

```python
acc_sc = accuracy_score(y_test, y_pred)
print(acc_sc)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
import numpy as np

X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 1, stop=X_set[:, 0].max() + 1, step=0.01),
    np.arange(start=X_set[:, 1].min() - 1, stop=X_set[:, 1].max() + 1, step=0.01)
)
```

```python
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

plt.contourf(
    X1, X2,
    model.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
    alpha=0.75,
    cmap=ListedColormap(('black', 'green'))

)

plt.scatter(
    X_set[:, 0], X_set[:, 1],
    c=y_set,
)

plt.title('Logistic Regression Decision Boundary (Training set)')
plt.xlabel('Age (scaled)')
plt.ylabel('Estimated Salary (scaled)')
plt.show()
```
