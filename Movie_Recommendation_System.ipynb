## Merge Datasets by "title"


```python
df = movies.merge(credits, on = 'title')
```

## Select Important Columns


```python
df = df[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
df.head()
```

## Handle Missing Values


```python
df.isnull().sum()
```

```python
# Remove rows with missing values
df.dropna(inplace=True)
```

## Convert Strings to Python List


```python
import ast

# Function to extract genre and keywords
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
df['genres'] = df['genres'].apply(convert)
df['keywords'] = df['keywords'].apply(convert)
```

```python
# Function to get first 3 cast members
def convert_cast(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L

df['cast'] = df['cast'].apply(convert_cast)
```

```python
# Function to get director name
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L

df['crew'] = df['crew'].apply(fetch_director)
```

```python
# Convert overview sentence into list of words
df['overview'] = df['overview'].apply(lambda x:x.split())
```

```python
# Remove spaces from Names

df['genres'] = df['genres'].apply(lambda x:[i.replace(" ","") for i in x])
df['keywords'] = df['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
df['cast'] = df['cast'].apply(lambda x:[i.replace(" ","") for i in x])
df['crew'] = df['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
# Combine all important columns to create new 'tags' column
df['tags'] = df['overview'] + df['genres'] + df['keywords'] + df['cast'] + df['crew']
```

```python
df
```

## Create new DataFrame


```python
new_df = df[['movie_id','title','tags']].copy()
```

```python
new_df
```

```python
# Convert Tags into string
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df
```

## Convert Text to Numerical Vectors


```python
from sklearn.feature_extraction.text import CountVectorizer

# Create CountVectorizer object
cv = CountVectorizer(max_features=5000, stop_words='english')

# Convert tags to vectors
vectors = cv.fit_transform(new_df['tags']).toarray()
```

## Calculate Similarity:

Cosine similarity measures how similar two movies are.


```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate cosine similarity between movies
similarity = cosine_similarity(vectors)
print(similarity)
```

## Recommendation Function


```python
# Function to recommend movies
def recommend(movie):

    # Get index of movie
    movie_index = new_df[new_df['title'] == movie].index[0]

    # Get similarity scores
    distances = similarity[movie_index]

    # Sort movies based on similarity score
    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    # Print recommended movies
    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Spider-Man 2")
```
